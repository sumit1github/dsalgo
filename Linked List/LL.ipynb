```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None
        
        #nnumber of elements
        self.n = 0

    def __len__(self):
        return self.n

    def add_in_head(self, data):
        '''
        - LL can/can't be empty
        - if empty 
            - just put data in the head
        - if not
            - need to add data in head
            - need to store the ref of prev head in to the new head's address
        '''
        new_node = Node(data)

        # if LL is empty
        if not self.head:
            self.head = new_node
            

        # not empty    
        else:
            temp = self.head
            self.head = new_node
            self.head.next = temp

        self.n += 1
    
    def traverse(self):
        current = self.head
        result = ""
        while current != None:
            result += f" {current.data} -->"
            current = current.next
        print(result[:-4])  # not to print last " -->"

    def insert_in_tail(self, data):
        '''
        - it is not possible to directly access tail
        - need to start from head
        - tail: who is pointing to None
        - in while loop it will check untill-unless current.next  = None
        '''
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return

        current = self.head
        while current.next is not None:
            
            current = current.next
        
        current.next = new_node
        self.n += 1

    def insert_in_middle(self, target_value, data):
        '''
        target_value : 
            - after which value we are going to add the new data
        - eg : let prev data is 1 -> 2 -> 3 -> 4, need add 33 after 2. so the new LL will be 1 -> 2 -> 33 -> 3 -> 4 

        steps:
            - check for the empty LL
            - run while loop till end
            - if the current.value == target_value
                - store the next node in temp variable
                - current.next will point to new_node
                - new node will point to node store in temp
        '''

        if not self.head:
            print("No such value present")
            return
        new_node = Node(data)
        current  = self.head

        while current.next is not None:
            if current.data == target_value:
                temp = current.next

                current.next = new_node

                new_node.next = temp
                return

            current = current.next
        self.n += 1

    def delete_head(self):
        '''
        it is just need need point to next node
        '''

        if self.head is not None and self.head.next is not None:
            self.head  = self.head.next

    def pop_last(self):
        '''
        delete data from the tail
        need to do : second_last_element.next = None
        step:
            - while loop
                - checking the current.next.next
                - if current.next.next === None
                    - current is the second last element
            - current.next = None
        '''

        #
        if self.head is not None and self.head.next is not None:
            current  = self.head

            while current.next.next is not None:
                current = current.next
                
            current.next = None

    def delete_by_value(self, target_value):
        # If head needs to be deleted
        if self.head and self.head.data == target_value:
            self.delete_head()
            return

        current = self.head
        while current and current.next:
            if current.next.data == target_value:
                current.next = current.next.next
                return  # Exit after deleting first match
            else:
                current = current.next

    def search(self, target_value):
        if not self.head:
            return None
        
        index = 0
        current = self.head
        while current is not None:
            
            if current.data == target_value:
                return index

            current = current.next
            index += 1

        return None
    
    def __getitem__(self, index_position):
        '''
        like array if we want to get value like data[2]==> value
        '''

        current  = self.head
        position = 0
        while current is not None:
            
            if position == index_position:
                return current.data

            current = current.next
            position += 1

        return "Index error"
    
    def max_value(self):

        if not self.head.next:
            return self.head

        current = self.head
        max_val = self.head.data

        while current != None:
            if current.data > max_val:
                max_val = current.data

            current = current.next
        return max_val
    
    def reverse(self):
        current = self.head
        prev_node = None

        while current is not None:
            next_node = current.next      # store next node
            current.next = prev_node      # reverse the link
            prev_node = current           # move prev forward
            current = next_node           # move current forward

        self.head = prev_node             # update head to the last node
```

```python
LL = LinkedList()
```

# adding data in head

```python
LL.add_in_head(11)
LL.add_in_head(2)
LL.add_in_head(3)
LL.add_in_head(31)
LL.add_in_head(32)
LL.add_in_head(33)
```

```python
print(LL.traverse())
```

# adding data in tail

```python
LL.insert_in_tail(10)
print(LL.traverse())
```

# add data after some value

```python
LL.insert_in_middle(2,33)
print(LL.traverse())
```

# delete head

```python
LL.delete_head()
print(LL.traverse())
```

```python
# pop

LL.pop_last()
print(LL.traverse())
```

```python

# delete by value
LL.delete_by_value(11)
print(LL.traverse())
```

```python
# find max value
LL = LinkedList()

LL.add_in_head(11)
LL.add_in_head(2)
LL.add_in_head(3)
LL.add_in_head(31)
LL.add_in_head(32)
LL.add_in_head(33)
LL.insert_in_tail(45)

print(LL.traverse())
LL.max_value()
```
