here we are going to create stack using LL.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None     # <==== same as self.head
        self.count = 0      # <==== no of elements present in the LL/Stack

    def is_empty(self):
        '''
        if the self.top is empty then no data in the stack
        '''

        return True if not self.top else False
    
    def peek(self):
        '''
        it is used for to see the top element
        '''
        return None if not self.top else self.top.data
    
    def size(self):
        return self.count
    
    def push(self, data):

        new_node = Node(data)

        if not self.top:
            self.top = new_node
            self.count += 1
            return

        old_head  = self.top
        self.top = new_node
        new_node.next = old_head
        self.count += 1

    def print_stack(self):
        
        current  = self.top

        while current is not None:

            print(f"{current.data}")

            current = current.next

    def pop(self):
        '''
        it will delete the data from the end.
        
        becase we are making stack using LL, so our last element is the head of the LL. Need to make second element as head

        eg:
            1 -> 2 -> 3 -> 4      [head is  = 1]

            after pop

            2 -> 3 -> 4           [head is  = 2]
        '''

        
        if not self.is_empty():             # if not empty => False
        
            old_head = self.top
            self.top = self.top.next
            self.count -= 1
            return old_head.data
        
        
        return True                         # if empty => True
```

### 1. Now you have a given a text "sumit" now you need to reverse it using stack only

input -> 'sumit'

output - > 'timus'

```python
def reverse_string(input_data):

    # first create a stack using loop

    str_input = input_data
    s = Stack()

    for i in str_input:
        s.push(i)

    # noe we are going to do the pop untill last node

    final = ""

    while s.is_empty() == False:

        final += s.pop()

    return final


print(reverse_string("sumit"))
```
