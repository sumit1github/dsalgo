here we are going to create stack using LL.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Stack:
    def __init__(self):
        self.top = None     # <==== same as self.head
        self.count = 0      # <==== no of elements present in the LL/Stack

    def is_empty(self):
        '''
        if the self.top is empty then no data in the stack
        '''

        return True if not self.top else False
    
    def peek(self):
        '''
        it is used for to see the top element
        '''
        return None if not self.top else self.top.data
    
    def size(self):
        return self.count
    
    def push(self, data):

        new_node = Node(data)

        if not self.top:
            self.top = new_node
            self.count += 1
            return

        old_head  = self.top
        self.top = new_node
        new_node.next = old_head
        self.count += 1

    def print_stack(self):
        
        current  = self.top

        while current is not None:

            print(f"{current.data}")

            current = current.next

    def pop(self):
        '''
        it will delete the data from the end.
        
        becase we are making stack using LL, so our last element is the head of the LL. Need to make second element as head

        eg:
            1 -> 2 -> 3 -> 4      [head is  = 1]

            after pop

            2 -> 3 -> 4           [head is  = 2]
        '''

        if not self.top or self.top.next is None :   # checking is no element or only one element
            return "No data"
        
        self.top = self.top.next
        self.count -= 1
```

```python
# push data into stack

s = Stack()
s.push(11)
s.push(12)
s.push(13)
s.push(14)
s.push(115)
s.push(110)


s.print_stack()
```

```python
# pop

s = Stack()
s.push(11)
s.push(12)
s.push(13)
s.push(14)
s.push(115)
s.push(110)


s.print_stack()
s.pop()
print(50*"=")
s.print_stack()
```

```python
# peek

s = Stack()
s.push(11)
s.push(12)
s.push(13)
s.push(14)
s.push(115)
s.push(110)

s.peek()
```

```python
# get the size

s = Stack()
s.push(11)
s.push(12)
s.push(13)
s.push(14)
s.push(115)
s.push(110)


s.pop()
s.pop()
s.size()
```
